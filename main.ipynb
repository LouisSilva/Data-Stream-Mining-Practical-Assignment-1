```python
# Make the output a .txt file like what the ETAI labs do, and make it report the features used, and all the shit that's inside the outputted config.json like model parameters

config = load_yaml_config()

# Load the datasets
df_raw_lisboa = load_data_from_json("Lisboa", config)
df_raw_porto = load_data_from_json("Porto", config)

# Clean the datasets
df_cleaned_lisboa = clean_dataset(config["cleaning"], df_raw_lisboa)
df_cleaned_porto = clean_dataset(config["cleaning"], df_raw_porto)

lisboa_stream = prepare_city_stream(config["preprocessing"], df_cleaned_lisboa)
porto_stream = prepare_city_stream(config["preprocessing"], df_cleaned_porto)
```
